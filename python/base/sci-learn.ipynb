特征化:将两列特征分别减去均值、除以标准差，转化为均值为0、方差为1的分布。

```python
import numpy as np
np.random.seed(42)
feartures=np.random.rand(3,4)
# .rand(3,4)
```

```python
print(feartures)
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
feartures_standard=scaler.fit_transform(feartures)
print(feartures_standard)
```
