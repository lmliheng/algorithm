讲一下range，range 生成等差数列。直接打印range(5)不是一个list，range(5)是可迭代对象 iterable
很多方法接受iteralbe对象，比如list，sum

```python
print(list(range(5)))

print(list(range(0,5,2)))

print(list(range(5,0,-2)))

try:
    # range类
    print(range(5))
except Exception as e:
    print("发送错误:"+e)

print(sum(range(4)))
```

python里没有map方法，可以使用列表推导式，原理就是一个for循环，支持嵌套

```python
arr=[1,2,3]
double_arr=[x*2 for x in arr]
print(double_arr)

# 嵌套
arr_2d=[[0 for _ in arr] for _ in arr]
print(arr_2d)

# 支持filter方法
new_arr=[x for x in arr if x>1]
print(new_arr)
```

list遍历取出元素和索引 使用enumerate()

```python
for index,val in enumerate(arr): # 前面是索引
    print(val,index)
```

排序实例方法,变异方法，返回None
key 接受一个函数，用于指定排序的依据。
(method) def sort(
 *,
 key: None = None,
 reverse: bool = False
) -> None
Sort the list in ascending order and return None.
The sort is in-place (i.e. the list itself is modified) and stable (i.e. the order of two equal elements is maintained).
If a key function is given, apply it once to each list item and sort them, ascending or descending, according to their function values.
The reverse flag can be set to sort in descending order.

```python
arr=[4,2,3,1]
# arr.sort() 等价于list.sort(key=None, reverse=False)
# arr.sort(key=None,reverse=True)
print(arr)
```

创建数组
使用列表表达式或者numpy

```python
a1=[1,2,3]
a2=[[1,2],[2,3]]
# 列表表达式
a3=[0 for _ in range(4)]
a4=[[0 for _ in range(4)] for _ in range(4)]
print(a1)
print(a2)
print(a3)
print(a4)
```

浅拷贝

```python
arr=[1,2,1]
arr1=arr
print(arr1,arr)
arr1[0]=2
print(arr1,arr)
# : 浅拷贝
arr2=arr[:]
arr2[0]=3
print(arr2,arr)

# copy() 浅拷贝
arr3=arr.copy()
arr3[0]=4
print(arr3,arr)
```
