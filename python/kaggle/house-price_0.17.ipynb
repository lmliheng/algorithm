```python
import pandas as pd
import numpy as np
train_df=pd.read_csv('house-price/train.csv')
test_df=pd.read_csv('house-price/test.csv')

y=np.log1p(train_df["SalePrice"])
train_df=train_df.drop(["Id","SalePrice"],axis=1)
test_df=test_df.drop(["Id"],axis=1)
print(train_df.head())
print(train_df.shape)
```

```python
cols = ["OverallQual", "GrLivArea", "GarageCars", "TotalBsmtSF", "YearBuilt"]

X_train = train_df[cols]
X_test = test_df[cols]
```

```python
print(X_train.info())
print(X_test.info())
```

```python
X_test["TotalBsmtSF"] = X_test["TotalBsmtSF"].fillna(0)
X_test["GarageCars"] = X_test["GarageCars"].fillna(0)
print(X_test.info())
```

训练Laoss

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import Lasso

model = Lasso(alpha=0.001)  # alpha 控制惩罚力度，先不管
model.fit(X_train, y)

pred_log = model.predict(X_test)  # 这是 log 后的预测
pred = np.expm1(pred_log)  # 还原成真实价格
```

```python
sub = pd.DataFrame({"Id": pd.read_csv("house-price/test.csv")["Id"], "SalePrice": pred})
sub.to_csv("submit.csv", index=False)
```
