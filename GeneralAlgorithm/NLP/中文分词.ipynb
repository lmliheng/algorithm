```python
import jieba

text = "我爱北京天安门"

words = jieba.lcut(text)  # 返回列表
print(words)
```

全模式（把所有可能的词都切出来）

```python
words = jieba.lcut(text, cut_all=True)
print(words)
# 输出: ['我', '爱', '北京', '天安', '天安门']
```

添加自定义词典，处理专业术语或人名时很有用

```python
# 方式一：添加单个词
jieba.add_word("深度学习")
jieba.add_word("张三")

# 方式二：加载自定义词典文件（每行一个词）
jieba.load_userdict("my_dict.txt")

text = "张三正在研究深度学习"
print(jieba.lcut(text))
```

词性标注

```python
import jieba.posseg as pseg

words = pseg.lcut("我爱北京天安门")
for word, flag in words:
    print(f"{word} -> {flag}")
```

关键词提取

```python
from jieba import analyse

text = "自然语言处理是人工智能领域的重要方向"
keywords = analyse.extract_tags(text, topK=3)
print(keywords)
# 输出: ['自然语言处理', '人工智能', '重要']
```
